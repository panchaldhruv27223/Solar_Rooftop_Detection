## CODE for : Data creation  

```python
import os 
from PIL import Image

input_root = "tiff"
output_root = "Data"

os.makedirs(output_root,exist_ok=True)

target_size = (1024,1024)
patch_size =  1024
original_size = 1500
offset = original_size - patch_size

crops = [
    (0,0),
    (offset, 0),
    (0, offset),
    (offset, offset)
]


for subfolder in os.listdir(input_root):

    input_subfolder = os.path.join(input_root, subfolder)
    output_subfolder = os.path.join(output_root,subfolder)

    if os.path.isdir(input_subfolder):

        os.makedirs(output_subfolder, exist_ok=True)

        if "label" in subfolder.lower():
            resample_mode = Image.NEAREST
        else:
            resample_mode = Image.BILINEAR


        for filename in os.listdir(input_subfolder):

            if filename.lower().endswith(".tiff") or filename.lower().endswith(".tif"):
                input_path = os.path.join(input_subfolder, filename)

                output_file_name = os.path.splitext(filename)[0]

                with Image.open(input_path) as img:

                    for idx, (x,y) in enumerate(crops):
                        
                        patch = img.crop((x,y, x + patch_size, y+ patch_size))
                        patch = patch.resize((patch_size, patch_size), resample = resample_mode)

                        output_file = f"{output_file_name}_patch_{idx+1}.png"

                        output_path = os.path.join(output_subfolder, output_file)

                        patch.save(output_path)
```
